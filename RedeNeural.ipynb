## Criando Perceptron

```python
#Entrada
input = [1,2,3]
print(input)

#Pesos
weight = [0.2, 0.4,0.6]
print(weight)

#Bias
bias = 2
print(bias)

#Neuronio
output = input[0]*weight[0] + input[1]*weight[1] + input[2]*weight[2] + bias
print(output)
```

## Camada com 3 Neurônios


```python
inputLayer = [1.0, 2.0, 3.0, 2.5]

weightLayer1 = [0.1, -0.4, 2.0, 0.7]
weightLayer2 = [0.3, 0.2, 0.5, 0.1]
weightLayer3 = [0.2, 0.4, 0.6, 0.8]

biasLayer1 = 0.5
biasLayer2 = 0.3
biasLayer3 = 0.2
```

```python
outputLayer = [
    #neronio1
    inputLayer[0]*weightLayer1[0] + inputLayer[1]*weightLayer1[1] + inputLayer[2]*weightLayer1[2] + inputLayer[3]*weightLayer1[3] + biasLayer1,

    #neronio2
    inputLayer[0]*weightLayer2[0] + inputLayer[1]*weightLayer2[1] + inputLayer[2]*weightLayer2[2] + inputLayer[3]*weightLayer2[3] + biasLayer2,

    #neronio3
    inputLayer[0]*weightLayer3[0] + inputLayer[1]*weightLayer3[1] + inputLayer[2]*weightLayer3[2] + inputLayer[3]*weightLayer3[3] + biasLayer3
]

print(outputLayer)
```
